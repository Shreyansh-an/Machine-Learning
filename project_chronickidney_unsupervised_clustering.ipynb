AGGLOMERATIVE CLUSTERING

```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import AgglomerativeClustering, KMeans,DBSCAN
from sklearn.metrics import silhouette_score
from sklearn.decomposition import PCA
import matplotlib.pyplot as plt
import seaborn as sns

Agg=df11.copy()
# Apply Agglomerative Clustering
agg_clustering = AgglomerativeClustering(n_clusters=2)  # Assuming 3 clusters for the iris dataset
y_agg = agg_clustering.fit_predict(Agg)

# Apply KMeans Clustering
kmeans = KMeans(n_clusters=2)
y_kmeans = kmeans.fit_predict(Agg)

```

Comparing Agglomerative and Kmeans

```python
pca = PCA(n_components=2)
X_pca = pca.fit_transform(Agg)

# Plotting results
plt.figure(figsize=(14, 6))

# Agglomerative Clustering Plot
plt.subplot(1, 2, 1)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y_agg, palette='Set1', s=100)
plt.title('Agglomerative Clustering on Iris Dataset (PCA-reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

# KMeans Clustering Plot
plt.subplot(1, 2, 2)
sns.scatterplot(x=X_pca[:, 0], y=X_pca[:, 1], hue=y_kmeans, palette='Set2', s=100)
plt.title('KMeans Clustering on Iris Dataset (PCA-reduced)')
plt.xlabel('Principal Component 1')
plt.ylabel('Principal Component 2')

plt.show()

# Compare cluster assignments with ground truth labels
print("Comparison with True Labels:")
print(f"Agglomerative Clustering labels:\n{y_agg}")
print(f"KMeans Clustering labels:\n{y_kmeans}")
print(f"True Labels:\n{y}")
```

```python
# Evaluate Clustering using Silhouette Score
print("Silhouette Score (Agglomerative):", silhouette_score(Agg, y_agg))
print("Silhouette Score (K-Means):", silhouette_score(Agg, y_kmeans))
```

```python
# Apply DBSCAN clustering
dbscan = DBSCAN(eps=320, min_samples=8)  # You may need to tune eps and min_samples for better results
y_dbscan = dbscan.fit_predict(Agg)

# Check the number of clusters formed (-1 indicates noise)
n_clusters = len(set(y_dbscan)) - (1 if -1 in y_dbscan else 0)
n_noise = list(y_dbscan).count(-1)
print(f"Number of clusters: {n_clusters}")
print(f"Number of noise points: {n_noise}")
```

```python
plt.figure(figsize=(8, 6))
plt.scatter(X_pca[:, 0], X_pca[:, 1], c=y_dbscan , cmap='viridis', marker='o', edgecolor='k' )
plt.title(f'DBSCAN Clustering\nClusters: {n_clusters}, Noise Points: {n_noise}')
plt.xlabel('PCA Component 1')
plt.ylabel('PCA Component 2')
plt.colorbar(label='Cluster Label')
plt.show()
```

```python
# Calculate Silhouette Score
if len(set(y_dbscan)) > 1:  # Silhouette Score is undefined for a single cluster
    silhouette = silhouette_score(Agg, y_dbscan)
else:
    silhouette = "Undefined (Single Cluster)"

print(f"Silhouette Score: {silhouette}")
```
